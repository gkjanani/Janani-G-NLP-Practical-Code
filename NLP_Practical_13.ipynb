```python
import heapq
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
```

```python
heuristic = {'A': 5, 'B': 4, 'C': 2, 'D': 3, 'E': 1, 'F': 0}
```

```python
def best_first_search(start, goal):
  visited = set()
  queue = []
  heapq.heappush(queue, (heuristic[start], start))
  path = []

  while queue:
    h, node = heapq.heappop(queue)
    if node not in visited:
      path.append(node)
      visited.add(node)
      if node == goal:
        return path # Return path when goal is found
      for neighbor in graph[node]: # Corrected typo
        if neighbor not in visited:
          heapq.heappush(queue, (heuristic[neighbor], neighbor))

  return path # Return path after loop if goal not found (or partial path if goal is last node added)
```

```python
print("Best-First Search Path:", best_first_search('A','F'))
```

```python
graph = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 5},
    'D': {},
    'E': {'F': 2},
    'F': {}
}
```

```python
heuristic = {
    'A': 5,
    'B': 4,
    'C': 2,
    'D': 3,
    'E': 1,
    'F': 0
}
```

```python
def a_star_search(start, goal):
    open_set = []
    heapq.heappush(open_set, (heuristic[start], 0, start, [start]))  # (f, g, node, path)

    visited = set()

    while open_set:
        f, g, node, path = heapq.heappop(open_set)

        if node == goal:
            return path, g

        if node not in visited:
            visited.add(node)

            for neighbor, cost in graph[node].items():
                if neighbor not in visited:
                    g_new = g + cost
                    f_new = g_new + heuristic[neighbor]
                    heapq.heappush(open_set, (f_new, g_new, neighbor, path + [neighbor]))

    return None
```

```python
path, cost = a_star_search('A', 'F')
print("A* Search Path:", path)
print("Total Cost:", cost)
```
